### 4. Variables hold values

![Variables](https://raw.githubusercontent.com/LNB-DH/BSSDH_2026_LLM_API_workshop/main/img/session_0/variables.png)

A **variable** is a name that points to a value. The equals sign in `name = value` means **assign this value to this name**. It does not ask whether two things are equal.

Variable names can contain letters, numbers, and underscores, but cannot begin with a number. Descriptive names such as `document_count` are easier to understand than vague names such as `x`.

`````python
researcher_name = "Anna"
document_count = 12
language = "English"

print(researcher_name)
print(document_count)
print(language)
`````

`````python
summary = f"{researcher_name} is working with {document_count} documents."
print(summary)
`````

Assigning a new value updates a variable. Python first calculates the expression on the right, then stores the result under the name on the left.

Running the following update cell more than once adds `1` each time. Rerun the earlier variable-creation cell whenever you want to reset `document_count` to `12`.

`````python
document_count = document_count + 1
print(document_count)
`````

### 5. Values and types

Every Python value has a **type**. The type tells Python what kind of value it is and which operations make sense.

### 6. Basic operations

### 10. Comparisons and conditions

![Conditions](https://raw.githubusercontent.com/LNB-DH/BSSDH_2026_LLM_API_workshop/main/img/session_0/conditions.png)

Comparisons ask questions that produce `True` or `False`:

- `==` equal to
- `!=` not equal to
- `<` less than
- `>` greater than
- `<=` less than or equal to
- `>=` greater than or equal to

Remember: one equals sign `=` assigns a value; two equals signs `==` compare values.

`````python
year = 1918

print(year == 1918)
print(year < 1900)
print(year >= 1900)
`````

An `if` statement chooses what to do based on a condition. A colon `:` begins the controlled block, and the indented lines belong to that block.

**Indentation is part of Python syntax.** Keep the same four-space indentation for lines that belong together.

`````python
year = 1918

if year < 1900:
    period = "before 1900"
else:
    period = "1900 or later"

print(period)
`````

### 11. Loops

### 12. Functions

![Functions](https://raw.githubusercontent.com/LNB-DH/BSSDH_2026_LLM_API_workshop/main/img/session_0/functions.png)

A **function** is a reusable block of code. `def` defines the function and its parameters. `return` sends a result back to the code that called the function, while `print()` only displays something for a person to read.

As with conditions and loops, the indented lines form the function body.

`````python
def count_words(text):
    words = text.split()
    return len(words)


print(count_words("This is a short text."))
print(count_words("This is another example with more words."))
`````

A function can call another function and use conditions. Read the next example from the inside out: `count_words(text)` creates a number, and the `if` block chooses a label for that number.

`````python
def label_text_length(text):
    word_count = count_words(text)

    if word_count < 5:
        return "very short"
    elif word_count < 12:
        return "short"
    else:
        return "long"


print(label_text_length("Small note."))
print(label_text_length("This paragraph contains enough words to get a different label."))
`````

#### Function exercise

`````python
def label_document_by_year(year):
    if year < 1800:
        return "early source"
    elif year < 1950:
        return "modern historical source"
    else:
        return "recent source"


print(label_document_by_year(1750))
print(label_document_by_year(1918))
print(label_document_by_year(2020))
`````

### 14. When something goes wrong

Red error output is information, not evidence that you have broken Colab. Begin with the **last line** of the error, which usually names the problem.

| Error name | Common beginner cause | First thing to check |
| --- | --- | --- |
| `NameError` | A variable name does not exist yet | Did you run the earlier defining cell? Is the spelling identical? |
| `SyntaxError` | Python cannot understand the written structure | Check quotation marks, parentheses, commas, and missing colons |
| `IndentationError` | Lines in a block do not align | Use consistent four-space indentation inside `if`, `for`, and `def` |
| `TypeError` | An operation received an unsuitable type | Check for quoted numbers or attempts to combine text and numbers directly |

A practical recovery sequence:

1. Read the final error line.
2. Check the named line and the line immediately above it.
3. Compare spelling, quotation marks, parentheses, colons, and indentation with the nearest working example.
4. Rerun the cell that creates any missing variable, then rerun the failing cell.
5. If the notebook state has become confusing, use the **Runtime** menu to restart the session and run cells again from the top.

Make one small correction at a time. This makes it easier to learn which change solved the problem.

### 15. Asking an LLM for help with a cell

An LLM can act as a tutor, debugger, or Markdown assistant, but it can also produce plausible-looking mistakes. Ask about one focused problem, understand the proposed change, and run it yourself.

- **Gemini in Colab:** if the feature is available for your account, use the Gemini spark icon to open the notebook's AI panel. It may have access to relevant notebook context, but you should still name the cell and goal clearly.
- **ChatGPT or another external chat:** the model cannot see this notebook unless you paste or upload the relevant material. Provide the complete cell and complete error, not only one suspicious line.

A strong request gives the environment, goal, exact evidence, and desired form of the answer.

#### Prompt template for a code error

```text
I am a Python beginner using Google Colab.

This cell should [describe the intended result].

COMPLETE CODE CELL:
[paste the complete cell]

COMPLETE ERROR MESSAGE:
[paste the complete error after removing any private information]

Explain the error in plain language. Then give me the smallest corrected
version of the complete cell and tell me what output I should expect.
Do not add packages, file access, network requests, or API keys.
```

If the explanation is still too technical, follow up with: “Explain each changed line without assuming previous programming experience.”

#### Prompt template for a Markdown cell

```text
Write the source for one Google Colab Markdown cell containing a heading,
a short paragraph, a bullet list, and a link. Use clear language for a
Digital Humanities student. Return the Markdown source inside a fenced
markdown block so I can copy the source rather than only seeing it rendered.
```

You can also paste your Markdown source and ask the model to identify why a heading, list, link, or code block is not rendering as intended.

#### Use generated help safely

- Never paste API keys, passwords, personal information, confidential collections, or unpublished material into an LLM unless your institution and the service's data policy permit it.
- Inspect generated code before running it. Be especially cautious with package installation, shell commands beginning with `!`, file deletion, network requests, and credential handling.
- Prefer the smallest possible change and test it on the small workshop example first.

### 16. Mini practice: a small humanities data workflow

![Mini Humanities Workflow](https://raw.githubusercontent.com/LNB-DH/BSSDH_2026_LLM_API_workshop/main/img/session_0/mini_humanities_workflow.png)

This self-contained example combines the main beginner ideas:

1. Store several source texts in a list.
2. Define functions for counting and labelling.
3. Loop through every text.
4. Store each result in a dictionary.
5. Add the dictionaries to a results list.
6. Print a readable summary.

Run the next three cells in order. The functions are defined again here deliberately, so the mini-workflow still works even after a fresh runtime restart.

`````python
source_texts = [
    "The archive contains letters from several families.",
    "This newspaper issue reports on trade, agriculture, and local politics.",
    "A short catalogue note."
]

print("Number of source texts:", len(source_texts))
`````

`````python
def count_words(text):
    return len(text.split())


def simple_length_label(word_count):
    if word_count < 5:
        return "very short"
    elif word_count < 10:
        return "short"
    else:
        return "long"


analysis_results = []

for text in source_texts:
    word_count = count_words(text)
    result = {
        "text": text,
        "word_count": word_count,
        "length_label": simple_length_label(word_count)
    }
    analysis_results.append(result)

print("Number of results:", len(analysis_results))
`````

`````python
for result in analysis_results:
    print(result["word_count"], "words", "-", result["length_label"], "-", result["text"])
`````
